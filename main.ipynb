```python
import pandas as pd
```

```python
df = pd.read_csv('./CUSTOMER_SULAWESI.csv')
```

```python
df['invoice_date'] = pd.to_datetime(df['invoice_date'])
df['year'] = df['invoice_date'].dt.year.astype('Int64')
df['month'] = df['invoice_date'].dt.month
```

```python
df_customer_monthly = (
    df
    .groupby(["customer_name", "year", "month"], as_index=False)
    .agg(
        total_revenue=("total_price", "sum"),
        total_invoice=("invoice_number", "nunique")
    )
)
```

```python
df_customer_monthly.nlargest(10, 'total_invoice')
```

```python
df_final_customer = (
    df_customer_monthly
    .groupby("customer_name", as_index=False)
    .agg(
        total_revenue=("total_revenue", "sum"),
        total_invoice=("total_invoice", "sum"),
        median_invoice_per_bulan=("total_invoice", "median"),
        median_revenue_per_bulan=("total_revenue", "median")
    )
)
```

```python
df_final_customer.nlargest(10, 'median_invoice_per_bulan')
```
